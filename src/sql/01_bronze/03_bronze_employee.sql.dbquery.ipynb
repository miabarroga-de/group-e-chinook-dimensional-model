```sql
-- Loads the Employee source file into the Raw layer.
-- The existing Raw table is not overwritten.

USE CATALOG chinook;
USE SCHEMA ge_raw;


-- Creates the Raw Employee table only when it does not yet exist.

CREATE TABLE IF NOT EXISTS chinook.ge_raw.employee_raw
USING DELTA
AS
SELECT *
FROM read_files(
    '/Volumes/workspace/default/ftw-b12/shared/week05/chinook_csv/Employee.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);
```
