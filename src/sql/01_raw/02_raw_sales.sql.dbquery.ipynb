```sql
-- Loads the Invoice and InvoiceLine source files into the Raw layer.
-- The existing Raw tables are not overwritten.

USE CATALOG chinook;
USE SCHEMA ge_raw;


-- Creates the Raw Invoice table only when it does not yet exist.

CREATE TABLE IF NOT EXISTS chinook.ge_raw.invoice_raw
USING DELTA
AS
SELECT *
FROM read_files(
    '/Volumes/workspace/default/ftw-b12/shared/week05/chinook_csv/Invoice.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);


-- Creates the Raw InvoiceLine table only when it does not yet exist.

CREATE TABLE IF NOT EXISTS chinook.ge_raw.invoice_line_raw
USING DELTA
AS
SELECT *
FROM read_files(
    '/Volumes/workspace/default/ftw-b12/shared/week05/chinook_csv/InvoiceLine.csv',
    format => 'csv',
    header => true,
    inferSchema => true
);
```
